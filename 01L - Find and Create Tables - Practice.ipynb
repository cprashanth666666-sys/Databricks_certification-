### Task 2: Preview the new CSV file

Use `read_files` to peek at the contents of `employees2.csv` before creating a table.

### Task 3: Create a Delta table from the new file

Use `CREATE TABLE AS SELECT` to create a table called **`new_employees`** from `employees2.csv`.

```python
%sql
-- TODO: Create a new Delta table called new_employees from employees2.csv
CREATE TABLE IF NOT EXISTS new_employees
AS SELECT * FROM read_files('/Volumes/dbacademy/get_started_de/myfiles/employees2.csv');
```
